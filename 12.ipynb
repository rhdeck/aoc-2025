# Data Loader

```python
import numpy as np
import re


with open("12.txt") as f:
    data = f.read().splitlines()
    boxes = []
    tests = []
    for i, line in enumerate(data):
        if line == "":
            continue
        if re.search(r"^\d+:.*", line):
            # print(data[i+1:i+5])
            lines = [[(1 if x == "#" else 0) for x in g] for g in[list(y) for y in data[i+1:i+5]]]
            # print(lines)
            lns = np.array([np.array(x) for x in lines[:-1]])
            # print(lns)
            boxes.append(lns)
        elif re.search(r"^\d+x\d+:.*", line):
            l = [x.strip(":").split("x") for x in line.split(" ")]
            # print(l)
            tests.append((int(l[0][0]), int(l[0][1]), [int(x[0]) for x in l[1:]]))
tests[0][2]
# boxes   
```

# Part 1

```python
simple_passes = 0
fails = 0
failScores = []
failTests = []
passTests = []
deadTests = []
# for box in boxes:
#     print(box)
# passScores = []
fills = [int(sum([sum(r) for r in box])) for box in boxes]
print("fills", fills)

for x,y, scores in tests:
    totalScore = sum(scores)
    limitX = x // 3
    limitY = y // 3

    f = sum([t[0] * t[1] for t in zip(scores, fills)])
    # print("fills", f)
    if(totalScore <= limitX * limitY):
        # print("Pass")
        simple_passes += 1
        passTests.append((x,y,scores))
        # passScores.append((totalScore * 9 - x*y, x * y, limitX * limitY * 9))
    elif(f >= x * y):
        deadTests.append((x,y,scores, f))
    elif(totalScore > limitX * limitY):
        # print("Fail")
        fails += 1
        failScores.append((totalScore * 9 - x*y, x * y))
        failTests.append((x,y,scores, totalScore, limitX * limitY))
print("passes, impossible fits, fails for further examination ", simple_passes, len(deadTests), fails)
print("passTests", passTests)
print("deadTests", deadTests)
print("failTests", failTests)
# print("failScores", sorted(failScores))
# print("passScores", sorted(passScores))
```
